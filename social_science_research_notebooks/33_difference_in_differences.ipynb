# Difference-in-Differences (DiD) In-Depth

Policy evaluation technique comparing treatment and control groups over time.

```
import pandas as pd
import statsmodels.formula.api as smf

# Synthetic Data
# 2 Groups (0=Control, 1=Treat), 2 Periods (0=Pre, 1=Post)
df = pd.DataFrame({
    'group': [0, 0, 1, 1, 0, 0, 1, 1]*10,
    'period': [0, 1, 0, 1, 0, 1, 0, 1]*10,
    'outcome': [10, 12, 10, 15, 11, 13, 11, 18]*10 # Treatment effect is (15-10) - (12-10) = 5 - 2 = 3 approx
})

df['treatment'] = df['group'] * df['period']

model = smf.ols("outcome ~ group + period + treatment", data=df).fit()
print(model.summary())
print(f"\nEstimated Treatment Effect: {model.params['treatment']:.2f}")
```
