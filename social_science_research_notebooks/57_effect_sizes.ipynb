# Effect Size Calculation In-Depth

Calculating Cohen's d, Hedges' g, and Eta-squared.

```
import pingouin as pg
import numpy as np

# Two independent groups
group1 = [4, 5, 6, 5, 4, 3, 5]
group2 = [6, 7, 8, 7, 6, 5, 7]

# Cohen's d
d = pg.compute_effsize(group1, group2, eftype='cohen')
print(f"Cohen's d: {d:.3f}")

# Hedges' g (better for small samples)
g = pg.compute_effsize(group1, group2, eftype='hedges')
print(f"Hedges' g: {g:.3f}")
```

```
# Eta-squared (from ANOVA)
# Using common language effect size for interpretation
cles = pg.compute_effsize(group1, group2, eftype='cles')
print(f"Common Language Effect Size: {cles:.3f} (Probability that a random score from Group 1 is > Group 2)")
```
