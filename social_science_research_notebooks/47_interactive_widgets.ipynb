# Interactive Widgets

Using ipywidgets for interactive data exploration.

```
import ipywidgets as widgets
from IPython.display import display
import pandas as pd
import numpy as np

df = pd.DataFrame(np.random.randn(10, 3), columns=['A', 'B', 'C'])

def filter_df(limit):
    print(df[df['A'] > limit])

slider = widgets.FloatSlider(value=0, min=-3, max=3, step=0.1, description='Filter A >')
widgets.interactive(filter_df, limit=slider)
```
