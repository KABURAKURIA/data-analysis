# Mediation Analysis with Bootstrapping In-Depth

A more robust method for testing indirect effects using Pingouin.

```
import pingouin as pg
import pandas as pd
import numpy as np

np.random.seed(123)
n = 100
X = np.random.randn(n)
M = 0.4 * X + np.random.randn(n)
Y = 0.5 * M + 0.2 * X + np.random.randn(n)
df = pd.DataFrame({'X': X, 'M': M, 'Y': Y})

# Run Mediation
stats = pg.mediation_analysis(data=df, x='X', m='M', y='Y', seed=42, n_boot=1000)
print(stats)
```

```
# Interpretation
# ACME = Average Causal Mediation Effect (Indirect)
# ADE = Average Direct Effect
# Prop. Mediated = ACME / Total Effect
print("\nCheck the 'pval' column for significance of the indirect effect (ACME).")
```
