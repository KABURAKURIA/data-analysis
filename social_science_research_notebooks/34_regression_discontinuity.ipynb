# Regression Discontinuity Design (RDD) In-Depth

Causal inference near a cutoff point.

```
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import statsmodels.formula.api as smf

# Synthetic Data: Scholarship given if Test Score > 50
score = np.random.normal(50, 10, 100)
treated = (score > 50).astype(int)
# Outcome increased by 10 if treated
outcome = 5 + 0.5 * score + 10 * treated + np.random.normal(0, 2, 100)

df = pd.DataFrame({'score': score, 'treated': treated, 'outcome': outcome})

# Center score at cutoff
df['score_centered'] = df['score'] - 50

# Plot
plt.scatter(df['score'], df['outcome'], c=df['treated'], cmap='coolwarm')
plt.axvline(50, color='black', linestyle='--')
plt.title('Regression Discontinuity')
plt.show()
```

```
# RDD Regression
model = smf.ols("outcome ~ score_centered + treated", data=df).fit()
print(model.summary())
```
