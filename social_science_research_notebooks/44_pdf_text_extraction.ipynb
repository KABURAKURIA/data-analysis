# PDF Text Extraction

Extracting text from PDF files using pypdf. We first create a dummy PDF.

```
from reportlab.pdfgen import canvas
from pypdf import PdfReader

# 1. Create a dummy PDF
pdf_file = "sample.pdf"
c = canvas.Canvas(pdf_file)
c.drawString(100, 750, "This is a sample PDF for text extraction.")
c.drawString(100, 730, "Social science research often involves document analysis.")
c.save()
print(f"Created {pdf_file}")
```

```
# 2. Read the PDF
reader = PdfReader(pdf_file)
text = ""
for page in reader.pages:
    text += page.extract_text() + "\n"

print("Extracted Text:")
print(text)
```
