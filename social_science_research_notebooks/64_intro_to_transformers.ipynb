# Introduction to Transformers for NLP

The Transformer architecture (e.g., BERT, GPT) has revolutionized Natural Language Processing (NLP). Unlike older methods (TF-IDF, simple RNNs), Transformers capture deep contextual relationships in text.

In this notebook, we will use the **Hugging Face `transformers` library** to perform state-of-the-art NLP tasks with just a few lines of code.

**Tasks:**
1.  **Sentiment Analysis**: Determining if a text is positive or negative.
2.  **Zero-Shot Classification**: Classifying text into categories the model hasn't explicitly seen during training.


```python
from transformers import pipeline
import pandas as pd

print("Loading Sentiment Analysis Pipeline...")
# Using a small, fast model for demonstration to avoid memory issues
sentiment_classifier = pipeline("sentiment-analysis", model="distilbert-base-uncased-finetuned-sst-2-english")
print("Model loaded.")
```

## 1. Sentiment Analysis

Let's test the model on some social science-related text.

```python
texts = [
    "The new policy has significantly improved community trust.",
    "Residents are frustrated by the lack of communication from the council.",
    "The results of the study were inconclusive and require further investigation."
]

results = sentiment_classifier(texts)

for text, res in zip(texts, results):
    print(f"Text: {text}")
    print(f"Label: {res['label']}, Score: {res['score']:.4f}\n")
```

## 2. Zero-Shot Classification

Zero-shot classification allows us to categorize text without training a specific model for those categories. The model uses its general language understanding to guess the best fit.

*Note: This may download a larger model.*

```python
try:
    print("Loading Zero-Shot Pipeline (this may take a moment)...")
    classifier = pipeline("zero-shot-classification", model="facebook/bart-large-mnli")
    
    sequence = "The government announced a new budget allocation for public schools and universities."
    candidate_labels = ["politics", "education", "health", "economy"]
    
    result = classifier(sequence, candidate_labels)
    
    print(f"\nText: {sequence}")
    print("Scores:")
    for label, score in zip(result['labels'], result['scores']):
        print(f"- {label}: {score:.4f}")
        
except Exception as e:
    print(f"Could not load Zero-Shot model (likely memory/download limit): {e}")
```
