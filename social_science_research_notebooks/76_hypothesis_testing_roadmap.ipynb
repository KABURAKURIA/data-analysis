## Decision Logic

1.  **Type of Variables?**
    *   **Numerical vs Numerical** -> Correlation (Pearson/Spearman)
    *   **Categorical vs Categorical** -> Chi-Square Test
    *   **Numerical vs Categorical** -> Compare Means (T-test, ANOVA)

2.  **Number of Groups?** (for Numerical vs Categorical)
    *   **2 Groups** -> T-Test
    *   **3+ Groups** -> ANOVA

3.  **Dependent or Independent?**
    *   **Independent** (Different people) -> Independent T-test / One-way ANOVA
    *   **Dependent** (Same people, pre/post) -> Paired T-test / Repeated Measures ANOVA

4.  **Parametric Assumptions Met?** (Normality, Equal Variance)
    *   **Yes** -> Parametric Tests (T-test, ANOVA)
    *   **No** -> Non-Parametric Tests (Mann-Whitney, Kruskal-Wallis, Wilcoxon)


```
def choose_test(var_type, n_groups=None, paired=False, parametric=True):
    if var_type == 'num_vs_num':
        return "Pearson Correlation (Parametric) or Spearman Correlation (Non-Parametric)"
    
    elif var_type == 'cat_vs_cat':
        return "Chi-Square Test of Independence"
    
    elif var_type == 'num_vs_cat':
        if n_groups == 2:
            if paired:
                return "Paired T-Test (Parametric) or Wilcoxon Signed-Rank Test (Non-Parametric)"
            else:
                return "Independent T-Test (Parametric) or Mann-Whitney U Test (Non-Parametric)"
        elif n_groups > 2:
            if paired:
                return "Repeated Measures ANOVA (Parametric) or Friedman Test (Non-Parametric)"
            else:
                return "One-Way ANOVA (Parametric) or Kruskal-Wallis Test (Non-Parametric)"
    
    return "Unknown configuration"

# Example Usage
print("1. Compare income (num) between Men and Women (cat, 2 groups, independent):")
print(f"   -> {choose_test('num_vs_cat', n_groups=2, paired=False)}")

print("\n2. Compare test scores (num) for same students before and after training (cat, 2 groups, paired):")
print(f"   -> {choose_test('num_vs_cat', n_groups=2, paired=True)}")
```
