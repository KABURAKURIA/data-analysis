```python
import whisper
from gtts import gTTS
import os
import shutil

# Check for ffmpeg
if not shutil.which('ffmpeg'):
    print("WARNING: ffmpeg is not installed. Whisper requires ffmpeg.")
    print("If on Linux/Mac, install it (e.g., 'apt install ffmpeg').")
    print("If on Windows, add it to PATH.")
```

## 1. Generate Dummy Audio
We use Google Text-to-Speech to create a sample audio file.

```python
text = "This is a test of the audio transcription system. Social science research often involves interviews."
tts = gTTS(text=text, lang='en')
audio_file = "sample_audio.mp3"
tts.save(audio_file)
print(f"Created {audio_file}")
```

## 2. Transcribe with Whisper
We load the 'base' model (small and fast) to transcribe the file.

```python
if shutil.which('ffmpeg'):
    model = whisper.load_model("base")
    result = model.transcribe(audio_file)
    print("Transcription:")
    print(result["text"])
else:
    print("Skipping transcription due to missing ffmpeg.")
```
