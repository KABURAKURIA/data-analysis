# Advanced Visualization with Plotly In-Depth

While `matplotlib` and `seaborn` are excellent for static publications, `plotly` allows for interactive data exploration (zooming, hovering, filtering).

In this notebook, we will create:
1.  Interactive Scatter Plots (with hover data)
2.  Interactive Line Charts
3.  Box Plots with points
4.  3D Scatter Plots
5.  Funnel Charts


```python
import plotly.express as px
import pandas as pd
import numpy as np

# Using built-in Gapminder dataset if available, else synthetic
try:
    df = px.data.gapminder()
    print("Loaded Gapminder dataset.")
except:
    # Fallback synthetic data
    print("Generating synthetic data...")
    df = pd.DataFrame({
        'year': np.tile(range(2000, 2021), 5),
        'country': np.repeat(['A', 'B', 'C', 'D', 'E'], 21),
        'gdpPercap': np.random.uniform(1000, 50000, 105),
        'lifeExp': np.random.uniform(50, 90, 105),
        'pop': np.random.randint(1000000, 100000000, 105),
        'continent': 'Synthetic'
    })
    
df.head()
```

## 1. Interactive Scatter Plot

We can map multiple variables to dimensions: x-axis, y-axis, size, color, and animation frame (time).

```python
fig = px.scatter(df, 
                 x="gdpPercap", 
                 y="lifeExp", 
                 animation_frame="year", 
                 animation_group="country",
                 size="pop", 
                 color="continent", 
                 hover_name="country",
                 log_x=True, 
                 size_max=55, 
                 range_x=[100,100000], 
                 range_y=[25,90],
                 title="Life Expectancy vs GDP per Capita (Interactive)")

fig.show()
```

## 2. Interactive Box Plot

Box plots in Plotly can easily show underlying data points.

```python
fig = px.box(df, 
             x="continent", 
             y="lifeExp", 
             points="all", # Show individual points
             color="continent",
             title="Distribution of Life Expectancy by Continent")
fig.show()
```

## 3. 3D Scatter Plot

Useful for exploring relationships between three continuous variables.

```python
# Filter for latest year
latest_year = df['year'].max()
df_latest = df[df['year'] == latest_year]

fig = px.scatter_3d(df_latest, 
                    x='gdpPercap', 
                    y='lifeExp', 
                    z='pop',
                    color='continent',
                    hover_name='country',
                    title=f'3D Analysis ({latest_year})')
fig.show()
```

## 4. Funnel Chart

Funnel charts are commonly used in marketing and sales to visualize stages in a process (e.g., website visitors -> buyers).

```python
# Synthetic Funnel Data
data_funnel = dict(
    number=[3900, 2740, 2060, 1100, 200],
    stage=["Website Visit", "Downloads", "Potential Customers", "Requested Price", "Invoice Sent"]
)
fig = px.funnel(data_funnel, x='number', y='stage', title="Sales Funnel")
fig.show()
```
