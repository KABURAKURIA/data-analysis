# Nested JSON Flattening In-Depth

Handling complex JSON structures often found in social media API data.

```
import pandas as pd
import json

# Complex Nested JSON
data = [
    {
        "id": 1,
        "user": {"name": "Alice", "location": "NY"},
        "posts": [{"id": 101, "likes": 5}, {"id": 102, "likes": 10}]
    },
    {
        "id": 2,
        "user": {"name": "Bob", "location": "LA"},
        "posts": []
    }
]

# pandas normalizes simple nesting
df = pd.json_normalize(data)
print("Normalized Level 1:")
print(df)

# Exploding the list of posts
df_posts = df.explode('posts')
print("\nExploded Posts:")
print(df_posts)
```
