# Mixed ANOVA (Split-Plot Design)

Analyzing designs with both within-subjects and between-subjects factors.

```
import pandas as pd
import numpy as np
import pingouin as pg
import seaborn as sns
import matplotlib.pyplot as plt

# Synthetic Data: 20 Subjects (10 Control, 10 Treatment) measured at 2 Timepoints (T1, T2)
np.random.seed(42)
n_per_group = 10
subjects = list(range(n_per_group * 2))
groups = ['Control'] * n_per_group + ['Treatment'] * n_per_group

# T1 scores similar, T2 scores higher for Treatment
t1_scores = np.random.normal(10, 2, 20)
t2_scores_control = np.random.normal(10, 2, 10)
t2_scores_treatment = np.random.normal(15, 2, 10)
t2_scores = np.concatenate([t2_scores_control, t2_scores_treatment])

# Format for Pingouin (Long Format)
df = pd.DataFrame({
    'Subject': np.repeat(subjects, 2),
    'Group': np.repeat(groups, 2),
    'Time': np.tile(['T1', 'T2'], 20),
    'Score': np.zeros(40)
})

# Assign scores
for i, sub in enumerate(subjects):
    df.loc[(df['Subject'] == sub) & (df['Time'] == 'T1'), 'Score'] = t1_scores[i]
    df.loc[(df['Subject'] == sub) & (df['Time'] == 'T2'), 'Score'] = t2_scores[i]

print(df.head())
```

```
# Run Mixed ANOVA
aov = pg.mixed_anova(data=df, dv='Score', within='Time', between='Group', subject='Subject')
print(aov)
```

```
# Interaction Plot
sns.pointplot(data=df, x='Time', y='Score', hue='Group', capsize=.1)
plt.title('Group x Time Interaction')
plt.show()
```
