# Plotting Interaction Effects In-Depth

Visualizing 2-way and 3-way interactions using Seaborn.

```
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np

# Synthetic Data: 2 Factors (A, B) and DV
df = pd.DataFrame({
    'FactorA': np.repeat(['Low', 'High'], 20),
    'FactorB': np.tile(np.repeat(['Cond1', 'Cond2'], 10), 2),
    'Score': [10, 12]*10 + [15, 25]*10 + np.random.randn(40)
})

# Interaction: Effect of A depends on B
# When A is Low, B makes small difference (10 vs 12)
# When A is High, B makes large difference (15 vs 25)
```

```
# 1. Point Plot
plt.figure(figsize=(8, 6))
sns.pointplot(data=df, x='FactorA', y='Score', hue='FactorB', dodge=True, markers=['o', 's'])
plt.title('Interaction Plot (PointPlot)')
plt.show()
```

```
# 2. Bar Plot
plt.figure(figsize=(8, 6))
sns.barplot(data=df, x='FactorA', y='Score', hue='FactorB')
plt.title('Interaction Plot (BarPlot)')
plt.show()
```
