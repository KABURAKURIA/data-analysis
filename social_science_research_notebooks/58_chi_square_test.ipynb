# Chi-Square Test of Independence

Analyzing the relationship between two categorical variables.

```
import pandas as pd
import pingouin as pg
import scipy.stats as stats

# Synthetic Data: Gender x Preference
data = {
    'Gender': ['Male']*30 + ['Female']*30,
    'Preference': ['A']*10 + ['B']*20 + ['A']*20 + ['B']*10
}
df = pd.DataFrame(data)

# Contingency Table
observed = pd.crosstab(df['Gender'], df['Preference'])
print("Observed Frequencies:")
print(observed)
```

```
# Chi-Square Test
expected, observed, stats = pg.chi2_independence(df, x='Gender', y='Preference')
print("\nTest Statistics:")
print(stats)
```
