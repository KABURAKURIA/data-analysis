# Moderation Analysis

Testing if the relationship between X and Y depends on Z (Interaction).

```
import pandas as pd
import numpy as np
import statsmodels.formula.api as smf
import matplotlib.pyplot as plt
import seaborn as sns

N = 100
X = np.random.normal(0, 1, N)
Z = np.random.choice([0, 1], N)
# Effect of X on Y is stronger when Z=1
Y = 0.5 * X + 1.0 * (X * Z) + np.random.normal(0, 1, N)

df = pd.DataFrame({'X': X, 'Z': Z, 'Y': Y})

model = smf.ols("Y ~ X * Z", data=df).fit()
print(model.summary())

# Plot
sns.lmplot(x='X', y='Y', hue='Z', data=df)
plt.title('Moderation Effect')
plt.show()
```
