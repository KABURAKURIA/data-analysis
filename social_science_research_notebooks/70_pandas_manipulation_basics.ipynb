```python

import pandas as pd
import numpy as np

# 1. Create a Synthetic Dataset
data = {
    'ParticipantID': range(1, 21),
    'Group': np.random.choice(['Control', 'Treatment'], 20),
    'Age': np.random.randint(18, 65, 20),
    'Income': np.random.normal(50000, 15000, 20).round(2),
    'Satisfaction_Score': np.random.randint(1, 8, 20) # 1-7 Scale
}

df = pd.DataFrame(data)

# Show the first 5 rows
df.head()
```

## 2. Inspecting Data

```python

# Summary statistics
print(df.describe())

# Check data types
print(df.info())
```

## 3. Filtering Data

Selecting subsets of data based on conditions.

```python

# Select only the Treatment group
treatment_df = df[df['Group'] == 'Treatment']
print(f"Treatment Group Count: {len(treatment_df)}")

# Select participants older than 30 with High Satisfaction (>5)
target_group = df[(df['Age'] > 30) & (df['Satisfaction_Score'] > 5)]
target_group.head()
```

## 4. GroupBy and Aggregation

Comparing groups (like Pivot Tables).

```python

# Compare average Income and Satisfaction by Group
group_stats = df.groupby('Group')[['Income', 'Satisfaction_Score']].mean()
print(group_stats)
```

## 5. Creating New Columns

```python

# Create a binary column: 'High_Income' if Income > median
median_inc = df['Income'].median()
df['High_Income'] = df['Income'] > median_inc
df.head()
```
