# Named Entity Recognition (NER) In-Depth

NER involves identifying and classifying named entities (people, organizations, locations, dates) in text. This notebook compares NLTK (rule-based/statistical) and SpaCy (model-based) approaches, and visualizes the results on a synthetic news article.

```
import spacy
from spacy import displacy
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import nltk

sns.set_style('whitegrid')
```

## 1. Synthetic Data
We create a realistic-looking news snippet involving multiple entities.

```
text = """
Apple Inc. is planning to open a new campus in Raleigh, North Carolina by 2026. 
CEO Tim Cook announced the $1 billion investment on Monday. 
The tech giant aims to hire 3,000 new employees in fields like Machine Learning and AI. 
Google and Amazon have also expanded their presence in the East Coast recently. 
The project was discussed in a meeting at the White House last week.
"""
print(text.strip())
```

## 2. NER with SpaCy
SpaCy is an industrial-strength NLP library. We load the small English model (`en_core_web_sm`).

```
# Load model
try:
    nlp = spacy.load("en_core_web_sm")
except OSError:
    # Fallback if model isn't found (local execution)
    import subprocess
    import sys
    subprocess.run([sys.executable, "-m", "spacy", "download", "en_core_web_sm"])
    nlp = spacy.load("en_core_web_sm")

doc = nlp(text)

# Extract entities
entities = []
for ent in doc.ents:
    entities.append((ent.text, ent.label_))

df_ents = pd.DataFrame(entities, columns=['Entity', 'Label'])
print(df_ents)
```

## 3. Visualizing Entities
SpaCy's `displacy` visualizer highlights entities directly in the text.

```
displacy.render(doc, style="ent", jupyter=True)
```

## 4. Entity Statistics
Let's count the frequency of each entity type.

```
plt.figure(figsize=(8, 5))
sns.countplot(data=df_ents, x='Label', palette='pastel')
plt.title('Frequency of Entity Types')
plt.show()
```

## 5. Explaining Labels
SpaCy can explain what a label means.

```
unique_labels = df_ents['Label'].unique()
for label in unique_labels:
    print(f"{label}: {spacy.explain(label)}")
```
