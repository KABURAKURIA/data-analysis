# Named Entity Recognition (NER)

Identifies people, organizations, and locations in text using SpaCy or NLTK (using NLTK here for simplicity without heavy models).

```
import nltk
nltk.download('punkt', quiet=True)
nltk.download('punkt_tab', quiet=True)
nltk.download('averaged_perceptron_tagger_eng', quiet=True)
nltk.download('averaged_perceptron_tagger', quiet=True)
nltk.download('maxent_ne_chunker', quiet=True)
nltk.download('maxent_ne_chunker_tab', quiet=True)
nltk.download('words', quiet=True)

text = "Apple is looking at buying U.K. startup for $1 billion. Tim Cook is the CEO."

tokens = nltk.word_tokenize(text)
tagged = nltk.pos_tag(tokens)
entities = nltk.chunk.ne_chunk(tagged)

print(entities)
```
