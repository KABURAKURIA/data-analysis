# Social Network Construction & Analysis In-Depth

This notebook moves beyond basic graph creation to explore complex network analysis techniques using real-world-like data. We will analyze the famous Zachary's Karate Club graph and simulate a larger scale-free network.

```
import networkx as nx
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np

# Use Zachary's Karate Club graph (a standard social network dataset)
G = nx.karate_club_graph()

print(f"Nodes: {G.number_of_nodes()}")
print(f"Edges: {G.number_of_edges()}")
```

## 1. Centrality Measures
Centrality helps us identify the most important nodes in the network.
*   **Degree Centrality:** Number of connections.
*   **Betweenness Centrality:** How often a node acts as a bridge along the shortest path between two other nodes.
*   **Eigenvector Centrality:** Influence of a node based on the influence of its neighbors.

```
degree_dict = nx.degree_centrality(G)
betweenness_dict = nx.betweenness_centrality(G)
eigenvector_dict = nx.eigenvector_centrality(G)

# Create a DataFrame to compare
nodes = list(G.nodes())
df_centrality = pd.DataFrame({
    'Node': nodes,
    'Degree': [degree_dict[n] for n in nodes],
    'Betweenness': [betweenness_dict[n] for n in nodes],
    'Eigenvector': [eigenvector_dict[n] for n in nodes]
})

print("Top 5 Influential Members (by Degree):")
print(df_centrality.sort_values(by='Degree', ascending=False).head())
```

## 2. Advanced Visualization
We will visualize the network, sizing nodes by their degree centrality and coloring them by their 'club' affiliation (which is stored in the node attributes).

```
plt.figure(figsize=(10, 8))

# Layout
pos = nx.spring_layout(G, seed=42)  # Seed for reproducibility

# Node sizes based on degree (scaled up for visibility)
node_sizes = [v * 3000 for v in degree_dict.values()]

# Node colors based on club attribute
club_color_map = {'Mr. Hi': 'lightblue', 'Officer': 'orange'}
node_colors = [club_color_map[G.nodes[n]['club']] for n in G.nodes()]

nx.draw(G, pos, 
        with_labels=True, 
        node_color=node_colors, 
        node_size=node_sizes, 
        edge_color='gray', 
        alpha=0.9)

plt.title("Zachary's Karate Club (Size = Degree, Color = Club)")
plt.show()
```

## 3. Community Detection
We can algorithmically detect communities (clusters) within the network. We'll use the Greedy Modularity algorithm.

```
from networkx.algorithms.community import greedy_modularity_communities

communities = list(greedy_modularity_communities(G))
print(f"Detected {len(communities)} communities.")

# Create a color map for communities
color_map = {}
colors = ['#FF9999', '#66B2FF', '#99FF99', '#FFCC99']
for i, community in enumerate(communities):
    for node in community:
        color_map[node] = colors[i % len(colors)]

node_colors_comm = [color_map[n] for n in G.nodes()]

plt.figure(figsize=(10, 8))
nx.draw(G, pos, 
        with_labels=True, 
        node_color=node_colors_comm, 
        node_size=node_sizes, 
        edge_color='gray')

plt.title("Community Detection (Greedy Modularity)")
plt.show()
```

## 4. Simulating Large Social Networks
Real social networks often follow a 'Scale-Free' power law distribution (many small nodes, few massive hubs). We can simulate this using the Barabási-Albert model.

```
# Generate a scale-free graph with 100 nodes
BA = nx.barabasi_albert_graph(n=100, m=2)

plt.figure(figsize=(10, 8))
nx.draw(BA, node_size=30, node_color='purple', alpha=0.6)
plt.title("Simulated Scale-Free Network (Barabási-Albert)")
plt.show()

# Plot Degree Distribution
degrees = [d for n, d in BA.degree()]
plt.figure(figsize=(8, 5))
plt.hist(degrees, bins=range(min(degrees), max(degrees) + 1, 1), color='purple', alpha=0.7)
plt.title("Degree Distribution")
plt.xlabel("Degree")
plt.ylabel("Frequency")
plt.show()
```
