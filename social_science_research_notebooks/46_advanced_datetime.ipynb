# Advanced Date and Time Handling

Handling timezones and calculating durations.

```
import pandas as pd

df = pd.DataFrame({
    'timestamp_utc': pd.to_datetime(['2023-01-01 12:00', '2023-01-02 14:30'])
})

# localize to UTC
df['utc'] = df['timestamp_utc'].dt.tz_localize('UTC')

# convert to US/Eastern
df['eastern'] = df['utc'].dt.tz_convert('US/Eastern')

print(df)
```
