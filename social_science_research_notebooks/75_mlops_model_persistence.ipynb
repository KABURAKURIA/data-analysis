# MLOps: Model Persistence In-Depth

Saving and loading Machine Learning models for production use using `joblib` and `pickle`.

```
import joblib
import pickle
import pandas as pd
from sklearn.ensemble import RandomForestClassifier
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score

# 1. Train a Model
data = load_iris()
X_train, X_test, y_train, y_test = train_test_split(data.data, data.target, test_size=0.2, random_state=42)

model = RandomForestClassifier(n_estimators=100)
model.fit(X_train, y_train)

print(f"Model Accuracy: {accuracy_score(y_test, model.predict(X_test)):.2f}")
```

## 2. Saving with Joblib
`joblib` is often more efficient for objects that carry large numpy arrays.

```
# Save
joblib.dump(model, 'rf_model.joblib')
print("Model saved as rf_model.joblib")

# Load
loaded_model_joblib = joblib.load('rf_model.joblib')
print(f"Loaded Model Accuracy: {accuracy_score(y_test, loaded_model_joblib.predict(X_test)):.2f}")
```

## 3. Saving with Pickle
`pickle` is the standard Python serialization module.

```
# Save
with open('rf_model.pkl', 'wb') as f:
    pickle.dump(model, f)
print("Model saved as rf_model.pkl")

# Load
with open('rf_model.pkl', 'rb') as f:
    loaded_model_pickle = pickle.load(f)
    
print(f"Loaded Model Accuracy: {accuracy_score(y_test, loaded_model_pickle.predict(X_test)):.2f}")
```
