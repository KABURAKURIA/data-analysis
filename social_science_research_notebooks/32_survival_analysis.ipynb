# Survival Analysis In-Depth

Analyzing time-to-event data (e.g., time until employment, customer churn) using Lifelines.

```
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from lifelines import KaplanMeierFitter

# Synthetic Data
# Duration in months, Event (1=Occurred, 0=Censored)
data = pd.DataFrame({
    'durations': [5, 6, 6, 2.5, 9.6, 1.4, 2.6, 3, 5, 4],
    'event_observed': [1, 0, 0, 1, 1, 1, 0, 1, 0, 1]
})

print(data)
```

```
kmf = KaplanMeierFitter()
kmf.fit(data['durations'], data['event_observed'])

kmf.plot_survival_function()
plt.title('Kaplan-Meier Survival Curve')
plt.xlabel('Time (Months)')
plt.ylabel('Survival Probability')
plt.show()
```
