## 1. Analysis / Reporting with Excel (Python Equivalents)

While Excel is a foundational tool for data analysts, Python's `pandas` library offers more powerful, reproducible alternatives. Here is how common Excel functions map to Pandas:

### Key Mappings:
*   **VLOOKUP / HLOOKUP** $\rightarrow$ `pd.merge()` (VLOOKUP) or `df.loc[]` / Transposing `df.T` (HLOOKUP)
*   **Pivot Tables** $\rightarrow$ `df.pivot_table()`
*   **IF** $\rightarrow$ `np.where()` or `df.apply(lambda x: ...)`
*   **SUM / AVERAGE / MIN / MAX** $\rightarrow$ `df.sum()`, `df.mean()`, `df.min()`, `df.max()`
*   **COUNT** $\rightarrow$ `df.count()` or `len(df)` or `df['col'].value_counts()`
*   **DATEDIF** $\rightarrow$ `(df['date2'] - df['date1']).dt.days`
*   **REPLACE / SUBSTITUTE** $\rightarrow$ `df['col'].replace()` or `df['col'].str.replace()`
*   **UPPER / LOWER / PROPER** $\rightarrow$ `df['col'].str.upper()`, `.str.lower()`, `.str.title()`
*   **CONCAT** $\rightarrow$ `df['col1'] + df['col2']` or `df['col1'].str.cat(df['col2'])`
*   **TRIM** $\rightarrow$ `df['col'].str.strip()`


```
import pandas as pd
import numpy as np

# Synthetic Data: Sales Records
data = {
    'Date_Start': pd.date_range(start='2023-01-01', periods=5, freq='D'),
    'Date_End': pd.date_range(start='2023-01-05', periods=5, freq='D'),
    'Product_ID': [101, 102, 101, 103, 102],
    'Sales': [250, 150, 300, 200, 160],
    'Region': [' North ', 'south', 'north', 'East', ' south '], # Messy text
    'Comment': ['Good sale', 'bad sale', 'good deal', 'bad deal', 'ok sale']
}
df_sales = pd.DataFrame(data)

# Synthetic Data: Product Names (for VLOOKUP equivalent)
products = {
    'Product_ID': [101, 102, 103],
    'Product_Name': ['Widget A', 'Widget B', 'Widget C']
}
df_products = pd.DataFrame(products)

print("--- VLOOKUP Equivalent (Merge) ---")
df_merged = pd.merge(df_sales, df_products, on='Product_ID', how='left')
print(df_merged[['Product_Name', 'Sales']].head())

print("\n--- COUNT Equivalent ---")
print(f"Total Rows: {len(df_merged)}")
print("Counts per Region:\n", df_merged['Region'].value_counts())

print("\n--- Text Cleaning (TRIM, PROPER/TITLE) ---")
df_merged['Region_Clean'] = df_merged['Region'].str.strip().str.title()
print(df_merged[['Region', 'Region_Clean']].head())

print("\n--- SUBSTITUTE Equivalent ---")
df_merged['Comment_Fixed'] = df_merged['Comment'].str.replace('bad', 'poor')
print(df_merged[['Comment', 'Comment_Fixed']].head())

print("\n--- DATEDIF Equivalent (Days) ---")
df_merged['Days_Diff'] = (df_merged['Date_End'] - df_merged['Date_Start']).dt.days
print(df_merged[['Date_Start', 'Date_End', 'Days_Diff']].head())
```
