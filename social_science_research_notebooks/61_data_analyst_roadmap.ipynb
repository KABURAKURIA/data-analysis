## 1. Analysis / Reporting with Excel (Python Equivalents)

While Excel is a foundational tool for data analysts (VLOOKUP, Pivot Tables, IF statements), Python's `pandas` library offers more powerful, reproducible alternatives.

### Key Mappings:
*   **VLOOKUP / HLOOKUP** $\rightarrow$ `pd.merge()` or `df.join()`
*   **Pivot Tables** $\rightarrow$ `df.pivot_table()`
*   **IF / DATEDIF** $\rightarrow$ `np.where()` or `pd.to_datetime()` operations
*   **SUM / AVERAGE** $\rightarrow$ `df.sum()`, `df.mean()`


```
import pandas as pd
import numpy as np

# Synthetic Data: Sales Records
data = {
    'Date': pd.date_range(start='2023-01-01', periods=10, freq='D'),
    'Product_ID': [101, 102, 101, 103, 102, 101, 103, 103, 102, 101],
    'Sales': [250, 150, 300, 200, 160, 310, 210, 190, 155, 290],
    'Region': ['North', 'South', 'North', 'East', 'South', 'North', 'East', 'East', 'South', 'North']
}
df_sales = pd.DataFrame(data)

# Synthetic Data: Product Names (for VLOOKUP equivalent)
products = {
    'Product_ID': [101, 102, 103],
    'Product_Name': ['Widget A', 'Widget B', 'Widget C']
}
df_products = pd.DataFrame(products)

print("--- VLOOKUP Equivalent (Merge) ---")
# Merging sales with product names
df_merged = pd.merge(df_sales, df_products, on='Product_ID', how='left')
print(df_merged[['Date', 'Product_Name', 'Sales']].head())

print("\n--- Pivot Table Equivalent ---")
# Sum of Sales by Region and Product Name
pivot = df_merged.pivot_table(values='Sales', index='Region', columns='Product_Name', aggfunc='sum', fill_value=0)
print(pivot)

print("\n--- IF / Logic Equivalent ---")
# Flag high sales (> 200)
df_merged['High_Performance'] = np.where(df_merged['Sales'] > 200, 'Yes', 'No')
print(df_merged[['Product_Name', 'Sales', 'High_Performance']].head())
```
