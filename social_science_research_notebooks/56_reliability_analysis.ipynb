# Reliability Analysis (Cronbach's Alpha) In-Depth

Assessing the internal consistency of a scale/survey.

```
import pandas as pd
import pingouin as pg
import numpy as np

# Synthetic Data: 5 items, 50 subjects
np.random.seed(42)
# True score
true_score = np.random.normal(3, 1, 50)
# Items correlate with true score
data = {
    'Item1': true_score + np.random.normal(0, 0.5, 50),
    'Item2': true_score + np.random.normal(0, 0.5, 50),
    'Item3': true_score + np.random.normal(0, 0.5, 50),
    'Item4': true_score + np.random.normal(0, 0.6, 50),
    'Item5': np.random.normal(3, 1, 50) # Unrelated item (bad item)
}
df = pd.DataFrame(data)

print(df.head())
```

```
# Calculate Alpha for Items 1-4
alpha, ci = pg.cronbach_alpha(df[['Item1', 'Item2', 'Item3', 'Item4']])
print(f"Cronbach's Alpha (Items 1-4): {alpha:.3f}")

# Calculate Alpha with bad item
alpha_bad, ci_bad = pg.cronbach_alpha(df)
print(f"Cronbach's Alpha (All Items): {alpha_bad:.3f}")
print("Note how the alpha drops when including the bad item.")
```
