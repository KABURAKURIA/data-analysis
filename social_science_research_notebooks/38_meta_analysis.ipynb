# Meta-Analysis Visualization In-Depth

Visualizing effect sizes from multiple studies using a Forest Plot.

```
import matplotlib.pyplot as plt
import pandas as pd

studies = ['Study A', 'Study B', 'Study C', 'Study D']
effect_sizes = [0.2, 0.4, 0.1, 0.3]
ci_lower = [0.1, 0.2, -0.1, 0.1]
ci_upper = [0.3, 0.6, 0.3, 0.5]

df = pd.DataFrame({'Study': studies, 'ES': effect_sizes, 'Lower': ci_lower, 'Upper': ci_upper})

plt.figure(figsize=(6, 4))
plt.errorbar(df['ES'], df.index, xerr=[df['ES']-df['Lower'], df['Upper']-df['ES']], fmt='o', color='black', capsize=5)
plt.yticks(df.index, df['Study'])
plt.axvline(0, color='gray', linestyle='--')
plt.xlabel('Effect Size')
plt.title('Forest Plot')
plt.show()
```
