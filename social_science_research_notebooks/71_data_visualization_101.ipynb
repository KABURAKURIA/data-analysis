```python

import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
import numpy as np

# Set the theme
sns.set_theme(style="whitegrid")

# Generate Synthetic Data
np.random.seed(42)
df = pd.DataFrame({
    'Education_Years': np.random.randint(12, 22, 100),
    'Salary': np.random.normal(60000, 20000, 100),
    'Gender': np.random.choice(['Male', 'Female', 'Non-binary'], 100)
})
# Add a relationship: Higher education -> Higher salary
df['Salary'] += df['Education_Years'] * 3000
```

## 1. Histogram (Distribution)

```python

plt.figure(figsize=(8, 5))
sns.histplot(df['Salary'], kde=True, color='skyblue')
plt.title('Distribution of Salaries')
plt.xlabel('Annual Salary ($)')
plt.show()
```

## 2. Boxplot (Group Comparison)

```python

plt.figure(figsize=(8, 5))
sns.boxplot(x='Gender', y='Salary', data=df)
plt.title('Salary Distribution by Gender')
plt.show()
```

## 3. Scatter Plot (Relationship)

```python

plt.figure(figsize=(8, 5))
sns.scatterplot(x='Education_Years', y='Salary', hue='Gender', data=df, s=100)
plt.title('Education vs. Salary')
plt.show()
```

## 4. Pie Chart (Proportions)

```python

plt.figure(figsize=(6, 6))
df['Gender'].value_counts().plot.pie(autopct='%1.1f%%', startangle=90, colors=sns.color_palette('pastel'))
plt.title('Gender Distribution')
plt.ylabel('') # Hide y-label
plt.show()
```
