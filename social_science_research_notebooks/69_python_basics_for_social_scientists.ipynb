## 2. Variables and Data Types

Variables are containers for storing data values.

```python

# Integer (e.g., age, count)
participant_age = 25

# Float (e.g., reaction time, income)
reaction_time_ms = 450.5

# String (e.g., name, response)
response_text = "Strongly Agree"

# Boolean (e.g., consent given?)
has_consented = True

print(f"Age: {participant_age}, Type: {type(participant_age)}")
print(f"Response: {response_text}, Type: {type(response_text)}")
```

## 3. Lists (Collections of Data)

Lists are ordered collections. In research, a list might represent a column of data.

```python

# A list of ages from 5 participants
ages = [22, 45, 30, 25, 60]

print(f"Original list: {ages}")

# Accessing elements (0-indexed)
print(f"First participant: {ages[0]}")

# Adding a new participant
ages.append(33)
print(f"Updated list: {ages}")

# Basic statistics
average_age = sum(ages) / len(ages)
print(f"Average Age: {average_age:.2f}")
```

## 4. Dictionaries (Key-Value Pairs)

Dictionaries are useful for representing a single participant's record.

```python

participant = {
    "id": 101,
    "age": 30,
    "condition": "Control",
    "score": 85
}

print(f"Participant ID: {participant['id']}")
print(f"Condition: {participant['condition']}")
```

## 5. Loops

Loops allow us to repeat actions. For example, converting all Likert scores to a 0-1 scale.

```python

raw_scores = [1, 2, 3, 4, 5]
normalized_scores = []

for score in raw_scores:
    # Normalize to 0-1 range: (x - min) / (max - min)
    norm = (score - 1) / (5 - 1)
    normalized_scores.append(norm)

print(f"Raw: {raw_scores}")
print(f"Normalized: {normalized_scores}")
```

## 6. List Comprehensions

A concise way to create lists. It's very 'Pythonic' and widely used in data analysis.

```python
# Traditional loop
ages = [22, 45, 30, 25, 60]
old_way = []
for age in ages:
    if age > 30:
        old_way.append(age)

# List Comprehension
# [expression for item in list if condition]
over_30 = [age for age in ages if age > 30]

print(f"Over 30: {over_30}")
```

## 7. Functions

Functions are reusable blocks of code.

```python

def calculate_bmi(weight_kg, height_m):
    return weight_kg / (height_m ** 2)

# Example usage
p1_bmi = calculate_bmi(70, 1.75)
print(f"Participant 1 BMI: {p1_bmi:.2f}")
```

## 8. Error Handling (try-except)

Data is often messy. Error handling allows your code to continue running even if it encounters a problem (like a missing value).

```python
data_values = [10, 20, 'missing', 40, 50]
clean_data = []

for val in data_values:
    try:
        # Try to divide by 2
        res = val / 2
        clean_data.append(res)
    except TypeError:
        # If 'val' is a string, this runs
        print(f"Skipping invalid value: {val}")
        clean_data.append(None)
        
print(f"Cleaned data: {clean_data}")
```
