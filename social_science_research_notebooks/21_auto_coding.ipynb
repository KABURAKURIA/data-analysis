# Automated Coding Assistant In-Depth

Keyword-based extraction to assist qualitative coding.

```
import pandas as pd

text_data = [
    "I am very happy with the service.",
    "The wait time was too long.",
    "Staff was friendly but the food was cold.",
    "Excellent experience overall."
]

keywords = {
    'Positive': ['happy', 'excellent', 'friendly', 'good'],
    'Negative': ['long', 'cold', 'bad', 'poor']
}

def auto_code(text):
    codes = []
    for category, words in keywords.items():
        if any(w in text.lower() for w in words):
            codes.append(category)
    return codes

df = pd.DataFrame(text_data, columns=['Text'])
df['Codes'] = df['Text'].apply(auto_code)
print(df)
```
