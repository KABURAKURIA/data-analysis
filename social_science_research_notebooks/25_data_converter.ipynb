# Data Format Converter In-Depth

Converting between JSON, CSV, and Excel formats.

```
import pandas as pd
import json

# Create Sample Data
data = [
    {'id': 1, 'name': 'Alice', 'score': 85},
    {'id': 2, 'name': 'Bob', 'score': 90},
    {'id': 3, 'name': 'Charlie', 'score': 78}
]

# Save as JSON
with open('sample_data.json', 'w') as f:
    json.dump(data, f)

# 1. JSON to DataFrame
df = pd.read_json('sample_data.json')
print('Loaded from JSON:')
print(df)
```

```
# 2. DataFrame to CSV
df.to_csv('sample_data.csv', index=False)
print('\nSaved to sample_data.csv')

# 3. DataFrame to Excel
# Requires openpyxl
try:
    df.to_excel('sample_data.xlsx', index=False)
    print('Saved to sample_data.xlsx')
except ImportError:
    print('openpyxl not installed, skipping Excel save')
```
