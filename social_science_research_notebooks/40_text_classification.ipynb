# Supervised Text Classification

Classifying text into predefined categories using TF-IDF and Linear SVM.

```
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.svm import LinearSVC
from sklearn.pipeline import Pipeline

# Training Data
X_train = ["Election results are out", "Stock market crashed", "New movie released", "President gave a speech", "Inflation is high"]
y_train = ["Politics", "Economy", "Entertainment", "Politics", "Economy"]

classifier = Pipeline([
    ('tfidf', TfidfVectorizer()),
    ('clf', LinearSVC())
])

classifier.fit(X_train, y_train)

# Predict
print(classifier.predict(["The prime minister visited the school", "Prices are rising"]))
```
