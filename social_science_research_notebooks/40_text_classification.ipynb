# Supervised Text Classification In-Depth

Classifying text into predefined categories using TF-IDF with Linear SVM and Naive Bayes.

```
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.svm import LinearSVC
from sklearn.naive_bayes import MultinomialNB
from sklearn.pipeline import Pipeline

# Training Data
X_train = ["Election results are out", "Stock market crashed", "New movie released", "President gave a speech", "Inflation is high"]
y_train = ["Politics", "Economy", "Entertainment", "Politics", "Economy"]

# --- 1. Linear SVM ---
classifier = Pipeline([
    ('tfidf', TfidfVectorizer()),
    ('clf', LinearSVC())
])

classifier.fit(X_train, y_train)

# Predict
print("SVM Predictions:", classifier.predict(["The prime minister visited the school", "Prices are rising"]))

# --- 2. Naive Bayes (MultinomialNB) ---
nb_classifier = Pipeline([
    ('tfidf', TfidfVectorizer()),
    ('clf', MultinomialNB())
])

nb_classifier.fit(X_train, y_train)
print("Naive Bayes Predictions:", nb_classifier.predict(["The prime minister visited the school", "Prices are rising"]))
```
