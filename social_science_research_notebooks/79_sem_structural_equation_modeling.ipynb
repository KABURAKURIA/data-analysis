```python
import pandas as pd
import numpy as np
import semopy
from semopy import Model
import matplotlib.pyplot as plt

print(f"Semopy version: {semopy.__version__}")
```

## 1. Load Data
We use the classic 'Political Democracy' dataset often used in SEM tutorials.

```python
# Load built-in example data from semopy examples if available, or create it.
# Using the Political Democracy dataset included in semopy documentation
from semopy.examples import political_democracy
data = political_democracy.get_data()
desc = political_democracy.get_model()

print(data.head())
```

## 2. Define Model
The model specifies latent variables and regressions.

*   `ind60` is measured by `x1`, `x2`, `x3`
*   `dem60` is measured by `y1`, `y2`, `y3`, `y4`
*   `dem65` is measured by `y5`, `y6`, `y7`, `y8`
*   Regressions: `dem60 ~ ind60`, `dem65 ~ ind60 + dem60`


```python
print(desc)
```

## 3. Fit Model
We estimate the parameters.

```python
model = Model(desc)
results = model.fit(data)
print(results)
```

## 4. Inspect Results
We examine the coefficients and p-values.

```python
ins = model.inspect()
print(ins)
```
