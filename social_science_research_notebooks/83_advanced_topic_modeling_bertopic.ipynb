# Advanced Topic Modeling with BERTopic

Latent Dirichlet Allocation (LDA) (covered in notebook 03) is the classic method for topic modeling. However, it relies on a "Bag-of-Words" approach, ignoring the semantic meaning of words.

**BERTopic** is a modern topic modeling technique that leverages **Transformers** (BERT embeddings) to understand the context of words. It works by:
1.  Converting documents to vectors (Embeddings).
2.  Reducing dimensionality (UMAP).
3.  Clustering vectors (HDBSCAN).
4.  Extracting keywords (c-TF-IDF).

```python
from bertopic import BERTopic
from faker import Faker
import pandas as pd
import numpy as np

fake = Faker()
np.random.seed(42)
print("Libraries loaded.")
```

## 1. Generate Synthetic Data

We create a dataset of 200 "Research Abstracts" related to Social Science and Technology.

```python
topics = [
    "The study examines the impact of social media on mental health.",
    "This research explores the relationship between income inequality and crime rates.",
    "We analyze the effect of artificial intelligence on the labor market.",
    "A qualitative study of urban development and gentrification.",
    "Survey results indicating voter preferences in the upcoming election."
]

# Generate variations
docs = []
for _ in range(200):
    base = np.random.choice(topics)
    # Add some random noise/sentences
    noise = fake.sentence()
    docs.append(f"{base} {noise}")

print(f"Generated {len(docs)} documents.")
print("Example:", docs[0])
```

## 2. Fit BERTopic Model

We initialize and fit the model. 
*Note: This may take a minute as it downloads the embedding model.*

```python
# language='english' will load a lightweight sentence-transformer
topic_model = BERTopic(language="english", calculate_probabilities=True, verbose=True)
topics, probs = topic_model.fit_transform(docs)

print("Model Fitted.")
```

## 3. View Topics

Let's see the most frequent topics found.

```python
freq = topic_model.get_topic_info(); 
print(freq.head(10))
```

```python
# Inspect words in Topic 0
print(topic_model.get_topic(0))
```

## 4. Visualization

BERTopic comes with built-in visualization tools based on Plotly.

```python
fig = topic_model.visualize_topics()
fig.show()
```
