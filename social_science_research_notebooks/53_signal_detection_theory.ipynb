# Signal Detection Theory (SDT)

Calculating sensitivity ($d'$) and response bias ($c$ or $\beta$).

```
import numpy as np
from scipy.stats import norm

def calculate_sdt(hits, misses, false_alarms, correct_rejections):
    # Adjust for 0/1 rates (log-linear approach or 1/2n correction)
    # Using simple 0.5 correction to avoid infinity
    hits += 0.5
    misses += 0.5
    false_alarms += 0.5
    correct_rejections += 0.5
    
    n_signal = hits + misses
    n_noise = false_alarms + correct_rejections
    
    hit_rate = hits / n_signal
    fa_rate = false_alarms / n_noise
    
    # d' (d-prime)
    d_prime = norm.ppf(hit_rate) - norm.ppf(fa_rate)
    
    # c (criterion)
    c = -0.5 * (norm.ppf(hit_rate) + norm.ppf(fa_rate))
    
    return hit_rate, fa_rate, d_prime, c

# Example Data
h = 40  # Hits
m = 10  # Misses
fa = 15 # False Alarms
cr = 35 # Correct Rejections

hr, far, d, c = calculate_sdt(h, m, fa, cr)

print(f"Hit Rate: {hr:.2f}")
print(f"FA Rate: {far:.2f}")
print(f"d' (Sensitivity): {d:.2f}")
print(f"c (Bias): {c:.2f}")

if c > 0:
    print("Conservative bias (more likely to say no)")
else:
    print("Liberal bias (more likely to say yes)")
```
