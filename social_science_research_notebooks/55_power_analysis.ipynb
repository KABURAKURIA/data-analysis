# Power Analysis

Calculating required sample size or achieved power using statsmodels and Pingouin.

```
from statsmodels.stats.power import TTestIndPower
import matplotlib.pyplot as plt
import numpy as np

# 1. Calculate Sample Size for Independent T-Test
effect_size = 0.5  # Cohen's d (Medium effect)
alpha = 0.05       # Significance level
power = 0.80       # Desired power

analysis = TTestIndPower()
n_per_group = analysis.solve_power(effect_size=effect_size, alpha=alpha, power=power)
print(f"Required sample size per group: {n_per_group:.2f} (Round up to {int(np.ceil(n_per_group))})")
```

```
# 2. Power Curve Plot
fig = analysis.plot_power(dep_var='nobs', nobs=np.arange(5, 100), effect_size=np.array([0.2, 0.5, 0.8]))
plt.title('Power Analysis for T-Test')
plt.show()
```
