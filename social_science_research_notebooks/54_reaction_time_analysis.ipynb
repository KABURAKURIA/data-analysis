# Reaction Time Analysis In-Depth

Preprocessing RT data: Outlier removal, transformations, and visualization.

```
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

# Synthetic RT data (Ex-Gaussian like: Normal + Exponential)
np.random.seed(42)
n = 1000
rt_data = np.random.normal(400, 50, n) + np.random.exponential(100, n)
df = pd.DataFrame({'RT': rt_data})

print(df.describe())
```

```
# 1. Visualize Distribution
plt.figure(figsize=(10, 4))
sns.histplot(df['RT'], kde=True)
plt.title('Reaction Time Distribution')
plt.xlabel('Time (ms)')
plt.show()
```

```
# 2. Outlier Removal (Mean +/- 2.5 SD)
mean_rt = df['RT'].mean()
sd_rt = df['RT'].std()
cutoff_lower = mean_rt - 2.5 * sd_rt
cutoff_upper = mean_rt + 2.5 * sd_rt

df_clean = df[(df['RT'] > cutoff_lower) & (df['RT'] < cutoff_upper)]
print(f"Removed {len(df) - len(df_clean)} outliers.")
```

```
# 3. Log Transformation (to normalize)
df_clean['log_RT'] = np.log(df_clean['RT'])

plt.figure(figsize=(10, 4))
sns.histplot(df_clean['log_RT'], kde=True)
plt.title('Log-Transformed RT')
plt.show()
```
