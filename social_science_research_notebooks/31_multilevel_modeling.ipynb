# Multilevel Modeling (Hierarchical Linear Models)

Analyzing data where observations are nested within groups (e.g., students within schools).

```
import pandas as pd
import numpy as np
import statsmodels.api as sm
import statsmodels.formula.api as smf

# Synthetic Data: Students nested in Schools
np.random.seed(42)
n_schools = 10
n_students_per_school = 30

school_effects = np.random.normal(0, 5, n_schools)
school_ids = np.repeat(range(n_schools), n_students_per_school)

student_SES = np.random.normal(0, 1, n_schools * n_students_per_school)
# Outcome = Intercept + SES_effect + School_Effect + Error
math_scores = 50 + 2 * student_SES + school_effects[school_ids] + np.random.normal(0, 2, n_schools * n_students_per_school)

df = pd.DataFrame({'School': school_ids, 'SES': student_SES, 'Math_Score': math_scores})
print(df.head())
```

```
# Fit MixedLM (Random Intercept Model)
md = smf.mixedlm("Math_Score ~ SES", df, groups=df["School"])
mdf = md.fit()
print(mdf.summary())
```
