```
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
warnings.filterwarnings('ignore')

# 1. Generate Synthetic Data
np.random.seed(42)
n = 500
data = {
    'Age': np.random.normal(35, 10, n),
    'Income': np.random.normal(50000, 15000, n),
    'Education_Level': np.random.choice(['High School', 'Bachelor', 'Master', 'PhD'], n),
    'Satisfaction': np.random.randint(1, 11, n),
    'Hours_Worked': np.random.normal(40, 5, n)
}
df = pd.DataFrame(data)

# Introduce Missing Values
df.loc[np.random.choice(df.index, 50), 'Income'] = np.nan
df.loc[np.random.choice(df.index, 20), 'Satisfaction'] = np.nan

# Introduce Outliers
df.loc[0, 'Income'] = 1000000
df.loc[1, 'Age'] = 120

df.head()
```

## 2. Data Inspection

```
print("--- Info ---")
df.info()

print("\n--- Describe ---")
display(df.describe())

print("\n--- Missing Values ---")
print(df.isnull().sum())
```

## 3. Missing Value Analysis

```
import seaborn as sns
plt.figure(figsize=(10, 6))
sns.heatmap(df.isnull(), cbar=False, cmap='viridis')
plt.title('Missing Data Heatmap')
plt.show()
```

## 4. Univariate Analysis
Understanding the distribution of single variables.

```
# Histogram for Income
plt.figure(figsize=(10, 5))
sns.histplot(df['Income'], kde=True)
plt.title('Income Distribution')
plt.show()

# Boxplot for Age (to see outliers)
plt.figure(figsize=(10, 5))
sns.boxplot(x=df['Age'])
plt.title('Age Boxplot')
plt.show()
```

## 5. Bivariate Analysis
Relationships between variables.

```
# Scatterplot: Income vs Satisfaction
plt.figure(figsize=(10, 6))
sns.scatterplot(x='Income', y='Satisfaction', hue='Education_Level', data=df)
plt.title('Income vs Satisfaction by Education')
plt.show()

# Correlation Matrix
plt.figure(figsize=(10, 8))
numeric_df = df.select_dtypes(include=[np.number])
sns.heatmap(numeric_df.corr(), annot=True, cmap='coolwarm')
plt.title('Correlation Matrix')
plt.show()
```
